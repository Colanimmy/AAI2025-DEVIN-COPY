```python
customer_message = "My internet connection keeps disconnecting every 10 minutes. I restarted my router, but the problem is still happening."

print("CUSTOMER MESSAGE:")
print(customer_message)
```

```python
prompt1 = """
Category: Technical
Reason: The customer is experiencing repeated internet disconnections even after restarting the router.
"""

print("PROMPT MESSAGE:")
print(prompt1)
```

```python
prompt2 = """
Missing Information:

Does the connection drop on all devices or only one specific device?
Are you using Wi-Fi, Ethernet, or both when the disconnections occur?
Do the modem or router status lights change or show any warnings when the connection drops?
"""

print("PROMPT MESSAGE 2:")
print(prompt2)
```

```python
prompt3 = """
Recommended Solution:

Check that all modem and router cables, including power and Ethernet cables, are securely connected and undamaged.
Power cycle the modem and router by unplugging them from power for about 30 seconds, then reconnecting them and allowing them to fully restart.
Test the connection on another device or, if possible, with an Ethernet cable to help determine whether the issue is limited to Wi-Fi or affects the entire connection.
If the disconnections continue, check your internet service provider for reported outages or contact their technical support to have the connection and equipment checked.
"""

print("PROMPT MESSAGE 3:")
print(prompt3)
```

```python
prompt4 = """
Escalate: YES
Reason: The issue should be escalated if the connection continues to disconnect after the recommended troubleshooting and requires assistance beyond basic troubleshooting.
"""

print("PROMPT MESSAGE 4:")
print(prompt4)
```
