PLAN:
The program will ask the user to enter the Power levels of three Destiny 2 weapons: Kinetic/Primary, Energy, and Power/Heavy.

It will store these values in a dictionary, calculate their average, and identify the weapons with the highest and lowest Power levels.

To prevent errors, the program will use try-except to reject non-numeric input and a conditional statement to reject negative numbers. If the user enters invalid input, the program will ask them to try again.

If multiple weapons share the highest or lowest Power level, the program will display all tied weapons.

```python
#ACTION:


weapons = {}

weapon_names = ["Kinetic/Primary", "Energy", "Power/Heavy"]

for weapon in weapon_names:
    while True:
        try:
            power = int(input(f"Enter {weapon} weapon Power level: "))

            if power < 0:
                print("Power level cannot be negative. Try again.")
                continue

            weapons[weapon] = power
            break

        except ValueError:
            print("Invalid input. Please enter a whole number.")

average = sum(weapons.values()) / len(weapons)

highest = max(weapons.values())
lowest = min(weapons.values())

print("\n=== Destiny 2 Weapon Power Analyzer ===")
print(f"Average Weapon Power: {average:.2f}")

print("\nHighest Power Weapon:")
for weapon, power in weapons.items():
    if power == highest:
        print(f"{weapon}: {power}")

print("\nLowest Power Weapon:")
for weapon, power in weapons.items():
    if power == lowest:
        print(f"{weapon}: {power}")
```

REFLECTION: The program uses sum() and len() to calculate the average and max() and min() to identify the highest and lowest Power levels.

The try-except block prevents non-numeric inputs from crashing the program. The while loop ensures that users must enter a valid value before proceeding.

One potential weakness is that the program accepts zero as a valid Power level. For this exercise, zero is permitted because the requirement specifies rejecting negative numbers, not zero.

The program also handles ties by displaying every weapon that shares the highest or lowest value.

```python
#FINAL/FIXED VERSION

# Destiny 2 Weapon Power Analyzer

weapons = {}

weapon_names = [
    "Kinetic/Primary",
    "Energy",
    "Power/Heavy"
]

# Collect and validate weapon Power levels
for weapon in weapon_names:
    while True:
        try:
            power = int(input(
                f"Enter {weapon} weapon Power level: "
            ))

            if power < 0:
                print("Error: Power level cannot be negative.")
                continue

            weapons[weapon] = power
            break

        except ValueError:
            print("Error: Please enter a valid whole number.")

# Calculate average, highest, and lowest
average = sum(weapons.values()) / len(weapons)

highest = max(weapons.values())
lowest = min(weapons.values())

# Display results
print("\n=== DESTINY 2 WEAPON POWER ANALYZER ===")

print("\nYour Weapons:")
for weapon, power in weapons.items():
    print(f"{weapon}: {power}")

print(f"\nAverage Weapon Power: {average:.2f}")

print("\nHighest Power Weapon(s):")
for weapon, power in weapons.items():
    if power == highest:
        print(f"{weapon}: {power}")

print("\nLowest Power Weapon(s):")
for weapon, power in weapons.items():
    if power == lowest:
        print(f"{weapon}: {power}")

print("\nWeapon Power analysis complete!")
```
